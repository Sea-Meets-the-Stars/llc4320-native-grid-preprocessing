# List LLC4320 dates for a global run config

Reads the per-timestep stores under `s3://dbof/LLC4320/` (named
`<YYYYMMDDTHH>.zarr`, see `get_raw_data`) and emits the dates in the
`data.date_iterations` format expected by a generate-global config
(`iterations.DATE_FMT` = `'%Y-%m-%d %H:%M:%S'`).

Paste the printed block into `configs/global/run/globals_for_cutouts_v2_1.yaml`.

```python
from datetime import datetime

import dbof.io.filesystems as filesystems
from dbof.global_dataset_creation.iterations import DATE_FMT  # canonical config date format

S3_ENDPOINT = "https://s3-west.nrp-nautilus.io"
SRC = "dbof/LLC4320"        # s3://dbof/LLC4320/<YYYYMMDDTHH>.zarr
STORE_FMT = "%Y%m%dT%H"    # LLC4320 per-timestep store naming (get_raw_data.py)

_, fs = filesystems.create_s3_filesystems(S3_ENDPOINT)  # sync filesystem

dates = []
for entry in fs.ls(SRC, detail=False):
    name = entry.rstrip("/").split("/")[-1]
    if not name.endswith(".zarr"):
        continue
    stem = name[: -len(".zarr")]
    try:
        dates.append(datetime.strptime(stem, STORE_FMT))
    except ValueError:
        continue  # skip non-timestep stores (e.g. grid.zarr)

dates = sorted(set(dates))
date_strings = [d.strftime(DATE_FMT) for d in dates]
print(f"{len(date_strings)} dates found")
date_strings[:5]
```

```python
# YAML block ready to paste into the config's `data:` section
print("data:")
print("  date_iterations:")
for s in date_strings:
    print(f"    - '{s}'")
```
